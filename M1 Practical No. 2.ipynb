```python
import pandas as pd
```

```python
df = pd.read_csv("Aviation safety.csv")
```

```python
print(df.head())
```

```python
# Select Numerical Column
aviation = df["avail_seat_km_per_week"]
```

```python
# Mean
print("Mean:", aviation.mean())
```

```python
# Median
print("Median:", aviation.median())
```

```python
# Mode
print("Mode:")
print(aviation.mode())
```

```python
# Minimum Value
print("Minimum:", aviation.min())
```

```python
# Maximum Value
print("Maximum:", aviation.max())
```

```python
# Range
print("Range:", aviation.max() - aviation.min())
```

```python
# Variance
print("Variance:", aviation.var())
```

```python
# Standard Deviation
print("Standard Deviation:", aviation.std())
```

```python
# Count
print("Count:", aviation.count())
```

```python
# Sum
print("Sum:", aviation.sum())
```

```python
# Statistical Summary
print("\nStatistical Summary:")
print(aviation.describe())
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.hist(aviation, bins=10, edgecolor='black')
plt.title("Histogram of Available Seat Kilometers per Week")
plt.xlabel("Available Seat Kilometers per Week")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.boxplot(aviation)
plt.title("Box Plot of Available Seat Kilometers per Week")
plt.ylabel("Available Seat Kilometers per Week")
plt.show()
```
